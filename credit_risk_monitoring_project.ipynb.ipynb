```python
selected_cols = ['id',
    'loan_amnt',
    'term',
    'int_rate',
    'installment',
    'grade',
    'sub_grade',
    'annual_inc',
    'loan_status',
    'dti',
    'fico_range_low',
    'fico_range_high',
    'revol_util',
    'pub_rec_bankruptcies',
    'purpose',
    'addr_state'
]

df_clean = df[selected_cols]

df_clean.head()
```

```python
df_clean.info()
```

```python
df_clean = df_clean.dropna()
```

```python
df_clean.shape
```

```python
df_clean['loan_status'].value_counts(normalize=True) * 100     #percentage distribution , checking portfolio health
```

```python
baseline = 12
df_clean.loc[:,'utilization_diff'] = df_clean['revol_util'] - baseline
```

```python
df_clean.head()
```

```python
df_clean[['revol_util', 'utilization_diff']].head()
```

```python
df_clean.loc[:, 'high_util_risk'] = df_clean['utilization_diff'].apply(lambda x: 1 if x > 30 else 0)
```

```python
df_clean['high_util_risk'].value_counts()
```

```python
df_clean[['revol_util', 'utilization_diff', 'high_util_risk']].head(10)
```

```python
df_clean.groupby('high_util_risk')['loan_status'].value_counts() #raw counts
```

```python
pd.crosstab(df_clean['high_util_risk'],df_clean['loan_status'],normalize='index') * 100     #standardized risk comparison, percentages.
```

```python
df_clean.loc[:, 'high_dti_risk'] = df_clean['dti'].apply(lambda x: 1 if x > 20 else 0)
```

```python
df_clean['high_dti_risk'].value_counts()
```

```python
pd.crosstab(df_clean['high_dti_risk'],df_clean['loan_status'],normalize='index') * 100
```

```python
#created multi risk indicator
df_clean.loc[:, 'total_risk_score'] = (df_clean['high_util_risk'] + df_clean['high_dti_risk'])
```

```python
df_clean['total_risk_score'].value_counts()
```

```python
pd.crosstab(df_clean['total_risk_score'],df_clean['loan_status'],normalize='index') * 100
```

```python
df_clean.shape
```

```python
#higher scores correspond to worse loan outcomes.
#As risk score is increasing , charged off increase , fully paid decreased, late loan payers increase.
```

```python
customers_df = df_clean[[
    'id',
    'annual_inc',
    'fico_range_low',
    'fico_range_high',
    'pub_rec_bankruptcies',
    'addr_state',
    'dti'
]]
```

```python
customers_df.head()
```

```python
loans_df = df_clean[[
    'id',
    'loan_amnt',
    'term',
    'int_rate',
    'installment',
    'grade',
    'sub_grade',
    'loan_status',
    'purpose'
]]
```

```python
loans_df.head()
```

```python
customers_df['id'].is_unique #since customer_id will act as primary key , every customer has a unique identifier.
```

```python
#checking relationship consistency
loans_df['id'].isin(customers_df['id']).all()
```

```python
#That means: every loan is correctly linked to a customer.
#This is called:referential integrity.
```

- primary key
- relational tables
- referential integrity

These are: core database architecture concepts.

```python
merged_df = customers_df.merge(loans_df,on='id',how='inner') #relational join operation
```

```python
merged_df.head()
```

SQLAlchemy is a Python library used for:

- connecting to databases
- writing SQL operations in Python
- moving DataFrames into SQL
- reading SQL tables into Pandas

psycopg2-binary- This is the PostgreSQL driver.

Python alone cannot directly “talk” to PostgreSQL.
psycopg2 acts like the translator/connector.

FLOW - Pandas  →  SQLAlchemy  →  psycopg2  →  PostgreSQL

```python
from sqlalchemy import create_engine
```

What is create_engine?- It creates a connection object between Python and a database.

Think of it like:
Python  ←→  Engine/Connector  ←→  PostgreSQL

```python
engine = create_engine(
    'postgresql://postgres:preetika123@localhost:5432/credit_risk_db'
)
```

```python
customers_df.to_sql(
    'customers',
    engine,
    if_exists='replace',
    index=False
)
```

```python
loans_df.to_sql(
    'loans',
    engine,
    if_exists='replace',
    index=False
)
```
